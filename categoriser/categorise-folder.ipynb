```python
import os
import re
import yaml

FRONTMATTER_RE = re.compile(r'(?s)^---\s*\n(.*?)\n---\s*(.*)$')
```

```python

def assign_category_from_folder(base_folder, category_folder_map, force=False):
    """
    Assign a category to all notes in a folder based on the folder name.

    Parameters:
    - base_folder: str, path containing subfolders with notes
    - category_folder_map: dict, category -> folder name
      e.g. {"DS": "data-science", "ML": "machine-learning"}
    - force: bool, overwrite existing category if True
    """
    # Reverse mapping: folder -> category
    folder_to_category = {v.lower(): k for k, v in category_folder_map.items()}

    for category, folder_name in category_folder_map.items():
        folder_path = os.path.join(base_folder, folder_name)
        if not os.path.exists(folder_path):
            print(f"Folder {folder_path} does not exist, skipping.")
            continue

        for root, _, files in os.walk(folder_path):
            for file in files:
                if not file.endswith(".md"):
                    continue
                filepath = os.path.join(root, file)
                with open(filepath, "r", encoding="utf-8") as f:
                    content = f.read()

                m = FRONTMATTER_RE.match(content)
                if not m:
                    continue
                yaml_text, body = m.groups()
                yaml_block = yaml.safe_load(yaml_text) or {}

                if force or "category" not in yaml_block or not yaml_block["category"]:
                    yaml_block["category"] = category
                    new_yaml = yaml.dump(yaml_block, sort_keys=False).strip()
                    new_content = f"---\n{new_yaml}\n---\n{body}"
                    with open(filepath, "w", encoding="utf-8") as f:
                        f.write(new_content)
                    print(f"Updated {filepath} -> {category}")
                else:
                    print(f"Skipped {filepath}, already has category {yaml_block['category']}")
```

```python
CATEGORY_TO_FOLDER = {
    "DS": "data-science",
    "ML": "machine-learning",
    "DE": "data-engineering",
    "STATISTICS": "statistics",
    "CS": "computer-science",
    "LANG": "natural-language",
    "DEVOPS": "devops",
    "PM": "project-management",
    "INDUSTRY": "industry",
    "DATA_ANALYSIS": "data-analysis",
    "NONE": "uncategorised",
    "DL": "deep-learning",
    "OTHER": "OTHER",
    "PAPER": "PAPER"
}
```

```python

assign_category_from_folder(r"categories", CATEGORY_TO_FOLDER, force=False)
```
