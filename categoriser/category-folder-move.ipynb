```python
import os
import shutil
import yaml
import re

FRONTMATTER_RE = re.compile(r'(?s)^---\s*\n(.*?)\n---\s*(.*)$')
```

```python

def move_notes_by_category(base_folder, category_folder_map):
    """
    Move notes into subfolders based on their category.
    
    Parameters:
    - base_folder: str, the folder containing notes
    - category_folder_map: dict, e.g.
        {"DS": "DS", "DATA_ANALYSIS": "Data Analysis", "NONE": "Uncategorized"}
    """
    for root, _, files in os.walk(base_folder):
        for file in files:
            if not file.endswith(".md"):
                continue

            filepath = os.path.join(root, file)
            with open(filepath, "r", encoding="utf-8") as f:
                content = f.read()

            m = FRONTMATTER_RE.match(content)
            if not m:
                continue  # skip files without YAML
            yaml_text, _ = m.groups()
            yaml_block = yaml.safe_load(yaml_text) or {}
            category = yaml_block.get("category", "NONE")

            # Determine target folder
            target_subfolder = category_folder_map.get(category, category)
            target_folder = os.path.join(base_folder, target_subfolder)
            os.makedirs(target_folder, exist_ok=True)

            # Move the file
            target_path = os.path.join(target_folder, file)
            if os.path.abspath(filepath) != os.path.abspath(target_path):
                shutil.move(filepath, target_path)
                print(f"Moved {filepath} -> {target_path}")
```

```python
CATEGORY_TO_FOLDER = {
    "DS": "data-science",
    "ML": "machine-learning",
    "DE": "data-engineering",
    "STATISTICS": "statistics",
    "CS": "computer-science",
    "LANG": "natural-language",
    "DEVOPS": "devops",
    "PM": "project-management",
    "INDUSTRY": "industry",
    "DATA_ANALYSIS": "data-analysis",
    "NONE": "uncategorised"
}
```

```python

move_notes_by_category("categories/uncategorised", CATEGORY_TO_FOLDER)
```
